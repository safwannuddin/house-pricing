# Handling Missing Values

The `total_bedrooms` column contains 207 missing (`NaN`) values.

Common ways to handle missing data:

1. Remove missing rows using `dropna()`  
   - Used when missing rows are very few and dataset is large.

2. Fill missing values with Mean  
   - Used when data is normally distributed and has fewer outliers.

3. Fill missing values with Median  
   - Best for skewed numerical data and datasets with outliers.

For this housing dataset, the **Median** method is used because housing data may contain outliers and median is more stable than mean.

```python
data["total_bedrooms"] = data["total_bedrooms"].fillna(
    data["total_bedrooms"].median()
)

````python
data["total_bedrooms"]=data["total_bedrooms"].fillna(
    data["total_bedrooms"].median()
)
````
